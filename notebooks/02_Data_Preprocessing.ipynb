# Load Dataset

```python
def load_dataset(folder: Path):

    datasets = {}

    csv_files = sorted(folder.glob("*.csv"))

    print(f"Found {len(csv_files)} files\n")

    for file in csv_files:

        try:

            df = pd.read_csv(file)

            datasets[file.stem] = df

            print(f"{file.name:<25} {df.shape}")

        except Exception as e:

            print(file.name, e)

    return datasets
```

```python
stock_data = load_dataset(RAW_STOCK_DIR)

index_data = load_dataset(RAW_INDEX_DIR)

etf_data = load_dataset(RAW_ETF_DIR)

commodity_data = load_dataset(RAW_COMMODITY_DIR)

forex_data = load_dataset(RAW_FOREX_DIR)

macro_data = load_dataset(RAW_MACRO_DIR)

news_data = load_dataset(RAW_NEWS_DIR)
```

## Standardize Column Names

```python
def clean_columns(df):

    df.columns = (
        df.columns
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )

    return df
```

```python
for datasets in [
    stock_data,
    index_data,
    etf_data,
    commodity_data,
    forex_data,
    macro_data,
    news_data
]:

    for key in datasets:

        datasets[key] = clean_columns(datasets[key])

print("Columns Standardized")
```

## Missing Value Analysis

Inspect missing values across all datasets before applying any cleaning strategy.

```python
def missing_value_report(datasets: dict):

    report = []

    for name, df in datasets.items():

        total_missing = df.isna().sum().sum()

        report.append({
            "Dataset": name,
            "Rows": len(df),
            "Columns": len(df.columns),
            "Missing Values": total_missing
        })

    return pd.DataFrame(report)
```

## Handle Missing Values

Apply appropriate missing value handling techniques while preserving the integrity of time-series data.

```python
def handle_missing_values(df: pd.DataFrame):

    df = df.copy()

    # Forward Fill
    df.ffill(inplace=True)

    # Backward Fill
    df.bfill(inplace=True)

    # Remove remaining missing values
    df.dropna(inplace=True)

    return df
```

```python
for datasets in [
    stock_data,
    index_data,
    etf_data,
    commodity_data,
    forex_data,
    macro_data
]:

    for key in datasets:

        datasets[key] = handle_missing_values(datasets[key])

print("Missing Values Handled Successfully")
```

## Duplicate Detection

```python
def duplicate_report(datasets):

    report = []

    for name, df in datasets.items():

        duplicates = df.duplicated().sum()

        report.append({
            "Dataset": name,
            "Duplicate Rows": duplicates
        })

    return pd.DataFrame(report)
```

```python
display(duplicate_report(stock_data))
```

## Remove Duplicate Records

```python
def remove_duplicates(df):

    df = df.drop_duplicates().reset_index(drop=True)

    return df
```

```python
for datasets in [
    stock_data,
    index_data,
    etf_data,
    commodity_data,
    forex_data,
    macro_data,
    news_data
]:

    for key in datasets:

        datasets[key] = remove_duplicates(datasets[key])

print("Duplicate Records Removed")
```

## Date Formatting

Convert date columns into datetime format and sort observations chronologically.

```python
def process_dates(df):

    if "Date" in df.columns:

        df["Date"] = pd.to_datetime(df["Date"])

        df.sort_values("Date", inplace=True)

        df.reset_index(drop=True, inplace=True)

    return df
```

```python
for datasets in [
    stock_data,
    index_data,
    etf_data,
    commodity_data,
    forex_data,
    macro_data
]:

    for key in datasets:

        datasets[key] = process_dates(datasets[key])

print("Dates Processed Successfully")
```

## Validate Market Data

Remove invalid observations before feature engineering.

```python
PRICE_COLUMNS = [
    "Open",
    "High",
    "Low",
    "Close",
    "Adj Close",
    "Adj_Close"
]

VOLUME_COLUMNS = [
    "Volume"
]


def validate_market_data(df):

    df = df.copy()

    for column in PRICE_COLUMNS:

        if column in df.columns:

            df[column] = pd.to_numeric(df[column], errors="coerce")

            df = df[df[column].notna()]

            df = df[df[column] > 0]

    for column in VOLUME_COLUMNS:

        if column in df.columns:

            df[column] = pd.to_numeric(df[column], errors="coerce")

            df = df[df[column].notna()]

            df = df[df[column] >= 0]

    return df.reset_index(drop=True)
```

```python
datasets = {
    "Stocks": stock_data,
    "Indices": index_data,
    "ETFs": etf_data,
    "Commodities": commodity_data,
    "Forex": forex_data
}

for dataset_name, dataset in datasets.items():

    print(f"\nValidating {dataset_name}...")

    for key in dataset:

        dataset[key] = validate_market_data(dataset[key])

print("\nMarket Data Validation Completed.")
```

## Outlier Detection

Detect potential outliers in numerical market data using the Interquartile Range (IQR) method.

Outliers are identified for analysis only and are **not removed** because extreme price movements can represent genuine market events.

```python
NUMERIC_COLUMNS = [
    "Open",
    "High",
    "Low",
    "Close",
    "Adj Close",
    "Adj_Close",
    "Volume"
]
```

```python
def detect_outliers(df):

    report = []

    for column in NUMERIC_COLUMNS:

        if column not in df.columns:
            continue

        series = pd.to_numeric(df[column], errors="coerce").dropna()

        if series.empty:
            continue

        q1 = series.quantile(0.25)
        q3 = series.quantile(0.75)

        iqr = q3 - q1

        lower = q1 - (1.5 * iqr)
        upper = q3 + (1.5 * iqr)

        outliers = series[(series < lower) | (series > upper)]

        report.append({
            "Feature": column,
            "Outliers": len(outliers),
            "Lower Bound": round(lower,2),
            "Upper Bound": round(upper,2)
        })

    return pd.DataFrame(report)
```

```python
for name, df in stock_data.items():

    print("="*80)

    print(name)

    display(detect_outliers(df))
```

## Data Integrity Check

Verify that the datasets satisfy the minimum requirements before feature engineering.

```python
def integrity_check(df):

    report = {

        "Rows": len(df),

        "Columns": len(df.columns),

        "Missing Values": df.isna().sum().sum(),

        "Duplicate Rows": df.duplicated().sum(),

        "Has Date": "Date" in df.columns,

        "Sorted": False

    }

    if "Date" in df.columns:

        report["Sorted"] = df["Date"].is_monotonic_increasing

    return report
```

```python
integrity_report = []

for name, df in stock_data.items():

    row = integrity_check(df)

    row["Dataset"] = name

    integrity_report.append(row)

integrity_report = pd.DataFrame(integrity_report)

display(integrity_report)
```

## News Dataset Cleaning

```python
def clean_news(df):

    df = df.copy()

    df.drop_duplicates(inplace=True)

    df.dropna(how="all", inplace=True)

    if "providerPublishTime" in df.columns:

        df["providerPublishTime"] = pd.to_datetime(
            df["providerPublishTime"],
            unit="s",
            errors="coerce"
        )

    return df.reset_index(drop=True)
```

```python
for key in news_data:

    news_data[key] = clean_news(news_data[key])

print("News datasets cleaned.")
```

## Macroeconomic Dataset Cleaning

```python
def clean_macro(df):

    df = df.copy()

    df.ffill(inplace=True)

    df.bfill(inplace=True)

    df.drop_duplicates(inplace=True)

    return df.reset_index(drop=True)
```

```python
for key in macro_data:

    macro_data[key] = clean_macro(macro_data[key])

print("Macroeconomic datasets cleaned.")
```

## Save Processed Datasets

```python
def save_datasets(datasets, output_dir):

    for name, df in datasets.items():

        output_path = output_dir / f"{name}.csv"

        df.to_csv(output_path, index=False)

    print(f"Saved -> {output_dir}")
```

```python
save_datasets(stock_data, PROCESSED_STOCK_DIR)

save_datasets(index_data, PROCESSED_INDEX_DIR)

save_datasets(etf_data, PROCESSED_ETF_DIR)

save_datasets(commodity_data, PROCESSED_COMMODITY_DIR)

save_datasets(forex_data, PROCESSED_FOREX_DIR)

save_datasets(macro_data, PROCESSED_MACRO_DIR)

save_datasets(news_data, PROCESSED_NEWS_DIR)
```
