# 07 - Investment Decision Engine & Risk Analytics
Model output is never a raw BUY label: LSTM signal + technicals + trend + volatility + macro + sentiment are combined into a weighted score -> BUY/HOLD/SELL with confidence, plain-language *why* and explicit *risks*.

> Logic lives in `src/engine/` (decision_engine.py, risk_analytics.py).

## 1. Decision for every covered symbol

```python
import pandas as pd

import src.api.service as svc

rows = []
for sym in svc.available_symbols():
    d = svc.decision_payload(sym)
    rows.append({'symbol': sym, 'rec': d['recommendation'], 'score': d['investment_score'],
                 'conf': d['confidence'], 'exp_ret': d['expected_return'], 'risk': d['risk_band']})
decisions = pd.DataFrame(rows).sort_values('score', ascending=False)
decisions
```

## 2. Full narrative for one symbol

```python
d = svc.decision_payload('NVDA')
print(d['recommendation'], f"| score {d['investment_score']} | confidence {d['confidence']:.0%}")
print('\nWHY:');   [print(' +', w) for w in d['why']]
print('\nRISKS:'); [print(' -', r) for r in d['risks']]
```

## 3. Risk analytics table

```python
risk_rows = {sym: svc.risk_payload(sym) for sym in svc.available_symbols()}
risk_table = pd.DataFrame(risk_rows).T
risk_table[['volatility','max_drawdown','sharpe','sortino','expected_return','risk_score','risk_band']]
```
